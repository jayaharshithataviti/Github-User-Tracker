```python
import requests
import pandas as pd

usernames = input("Enter github usernames separated by commas: ")

user_list = usernames.split(",")

all_users = []

for username in user_list:

    username = username.strip()

    url = f"https://api.github.com/users/{username}"

    response = requests.get(url)

    data = response.json()

    if response.status_code == 200:

        print("Username:", data["login"])

        print(f"{data['login']} has {data['followers']} followers")

        print(f"{data['login']} is following {data['following']} members")

        print(f"{data['login']} has {data['public_repos']} repositories")

        print(f"{data['login']} account was created at {data['created_at']}")

        print(f"{data['login']} account was updated at {data['updated_at']}")

        print(f"{data['login']} account bio is {data.get('bio', 'No bio available')}")

        print(f"{data['login']} lives in {data.get('location', 'No location available')}")

        print(f"{data['login']} company is {data.get('company', 'No company available')}")

        print(f"{data['login']} blog is {data.get('blog', 'No blog available')}")

        print(f"{data['login']} has {data.get('public_gists', 'No public gists available')} public gists")

        print(f"{data['login']} account type is {data['type']}")

        print(f"{data['login']} avatar URL is {data['avatar_url']}")

        followers = data["followers"]

        following = data["following"]

        if following != 0:
            ratio = followers / following
        else:
            ratio = followers

        print("Follower Ratio:", ratio)

        all_users.append({
            "Username": data["login"],
            "Followers": followers,
            "Following": following,
            "Repos": data["public_repos"],
            "Location": data.get("location", "Not available"),
            "Account Type": data["type"],
            "Follower Ratio": ratio
        })

    else:
        print(f"{username} not found")

df = pd.DataFrame(all_users)

print(df)

df.to_csv("github_users_data.csv", index=False)

print("CSV file saved successfully!")
    
```
